# univariate continous data----histrogram

```python
plt.figure(figsize=(10,5))
plt.hist(x="mpg",data=mt_cars_df)
plt.xlabel("MPG")
plt.ylabel("freq")
plt.title("MPG Distribution",size=15)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(x="disp",data=mt_cars_df)
plt.xlabel("disp")
plt.ylabel("Freg")
plt.title("disp Distribution",size=15)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(x="cyl" ,data=mt_cars_df)
plt.xlabel("cyl")
plt.ylabel("Freg") 
plt.title("cyl distribution")
plt.show()           


        
```

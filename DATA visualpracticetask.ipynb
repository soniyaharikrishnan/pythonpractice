# - Univariate Continous data---Histogram

```python
plt.figure(figsize=(10,5))
plt.hist(x="Daily_Change_Percent" ,data=finance_df)
plt.xlabel("Daily_Change_Percent")
plt.ylabel("freq")
plt.title("daily disribution",size=15)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(x="Commodity_Index" ,data=finance_df)
plt.xlabel("Commodity_Index")
plt.ylabel("freq")
plt.title("Commodity distribution",size=15)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(x="Bond_Yield_10Y_Percent" ,data=finance_df)
plt.xlabel("Bond_Yield_10Y_Percent")
plt.ylabel("freq")
plt.title("Bond distribution",size=15)
plt.show()
```

```python
plt.boxplot(x="Daily_Change_Percent",data=finance_df)
```
